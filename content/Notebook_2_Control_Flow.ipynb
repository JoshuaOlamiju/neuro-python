## What a Boolean really is

**The light switch.** Every light switch you have ever operated has exactly two
positions: on or off. There is no "a bit on." The position of the switch is a
Boolean: a value that is either `True` or `False`, nothing else.

**A pregnancy test.** The result strip reads positive or negative. However
complicated the underlying chemistry, the readout collapses to one of two
values. That readout is a Boolean.

In Python, a **Boolean** (named after the Victorian logician George Boole, who
formalised the mathematics of yes-and-no) is a value of type `bool`. It can
only ever be `True` or `False`, both spelled with a capital first letter, no
quotes. The moment you ask Python a yes-or-no question, the answer you get back
is a `bool`.

In neuroscience, the simplest question a model asks is: *did the neuron fire?*
A neuron (a brain cell that sends electrical signals) either crossed a voltage
threshold and fired a spike (an electrical pulse), or it did not. That
yes-or-no is a Boolean.

*(Voltage is the electrical "pressure" pushing charged particles across the
neuron's membrane. It is measured in millivolts, mV. The typical resting value
is about -70 mV, meaning the inside of the cell is 70 thousandths of a volt
more negative than the outside.)*


````
# A bool holds one of exactly two values
neuron_fired = True
below_threshold = False

print(neuron_fired)       # True
print(type(neuron_fired)) # <class 'bool'>

# Behind the scenes, True == 1 and False == 0
# This means you can count booleans like votes
spike_on_trial_1 = True
spike_on_trial_2 = False
spike_on_trial_3 = True

total_spikes = spike_on_trial_1 + spike_on_trial_2 + spike_on_trial_3
print("Total spikes:", total_spikes)  # 2 -- because True + False + True = 1 + 0 + 1
````

## Comparison operators: asking Python a yes-or-no question

A **comparison operator** is a symbol that compares two values and hands back
a `bool`. Think of it as writing a question using symbols.

**The waiting room.** The receptionist checks "Is this patient older than 65?"
and marks the form Yes or No. That mark is the result of a comparison.

**Your bathroom scale.** It tells you whether today's weight is above or below
your target. The display might show numbers, but the question "above target?"
has a Yes/No answer.

| Symbol | Reads as | Example | Result |
|--------|----------|---------|--------|
| `==` | equals | `3 == 3` | `True` |
| `!=` | does not equal | `3 != 4` | `True` |
| `<` | less than | `2 < 5` | `True` |
| `>` | greater than | `5 > 2` | `True` |
| `<=` | less than or equal | `3 <= 3` | `True` |
| `>=` | greater than or equal | `4 >= 5` | `False` |

**Critical distinction:** `=` (one equals) is the label gun -- it sticks a name
on a value. `==` (two equals) is a question -- it asks "are these the same?"
and returns `True` or `False`. Mixing them up is the single most common mistake
in this item.


````
# Membrane voltage example
# A neuron fires (produces a spike) when voltage crosses about -55 mV
voltage = -50.0          # current membrane voltage in millivolts
threshold = -55.0        # the firing threshold

is_above_threshold = voltage > threshold  # asks "is -50 greater than -55?"
print("Is above threshold:", is_above_threshold)  # True, because -50 > -55

# Checking equality -- did two neurons fire the same number of spikes?
spikes_neuron_A = 12
spikes_neuron_B = 12
same_count = spikes_neuron_A == spikes_neuron_B
print("Same spike count:", same_count)  # True

# not-equal
different = spikes_neuron_A != 8
print("Different from 8:", different)  # True

# Every comparison returns a bool you can store or use directly
print(type(is_above_threshold))  # <class 'bool'>
````

## Logical operators: combining yes-or-no questions

Once you can form one yes-or-no question, you need to combine them. Three words
do this: `and`, `or`, `not`.

**Triage in A&E.** "This patient needs immediate attention if their systolic BP
is below 90 AND they are unconscious." Both conditions must be true. That is
`and`.

**The alarm system.** "Trigger if motion is detected OR if the door opens." One
of the two is enough. That is `or`.

**The exemption form.** "Mark exempt if NOT registered." Flip the result. That
is `not`.

| Operator | Returns `True` when... |
|----------|----------------------|
| `A and B` | both A and B are `True` |
| `A or B` | at least one of A, B is `True` |
| `not A` | A is `False` |


````
# Comp-neuro scenario: classify a neuron's behaviour
voltage = -48.0      # mV
firing_rate = 80.0   # Hz (hertz: spikes per second -- a measure of how fast a neuron fires)

threshold = -55.0
high_rate_cutoff = 50.0

# and: both must be true
is_active = voltage > threshold and firing_rate > high_rate_cutoff
print("Neuron is active and high-firing:", is_active)   # True

# or: at least one must be true
needs_attention = voltage > -40.0 or firing_rate > 100.0
print("Needs attention:", needs_attention)   # False (neither condition met)

# not: flip the result
is_silent = not is_active
print("Neuron is silent:", is_silent)   # False

# Combining all three -- "excited but not saturated"
saturation_rate = 150.0
is_excited_but_not_saturated = is_active and not (firing_rate > saturation_rate)
print("Excited but not saturated:", is_excited_but_not_saturated)  # True
````

>
> **Short-circuit evaluation.** Python is lazy in a useful way. When evaluating
> `A and B`, if A is already `False`, Python does not even look at B -- the
> result must be `False` regardless. With `A or B`, if A is `True`, Python
> skips B entirely. This is called **short-circuit evaluation**. It matters in
> two real situations:
>
> 1. **Safety.** You can write `data is not None and data > 0` and Python will
>    not crash on the second check if `data` is `None`, because the first
>    condition fails and the rest is skipped.
> 2. **Speed.** Put the cheapest or most likely condition first; Python may avoid
>    doing expensive work on the right-hand side.
>
> **Operator precedence.** When you mix operators without parentheses, Python
> applies them in this order: `not` first, then `and`, then `or`. So
> `True or False and False` is `True` (because `False and False` is resolved
> first to `False`, then `True or False` is `True`). Parentheses always
> override precedence and make intent clear. When in doubt, use parentheses.
>
> **Chained comparisons.** Python allows `10 < x < 20`, which reads naturally
> as "x is between 10 and 20." Most languages do not allow this; Python does,
> and it means exactly what it looks like.
>
> **Truthiness.** Any value can be tested as a bool, not just actual bools. In
> a decision context, Python treats `0`, `0.0`, empty text `""`, and empty
> collections as `False`. Everything else is `True`. This is called
> **truthiness**. It means `if spike_list:` is a valid way to ask "does this
> list have anything in it?"
>
> **`is` vs `==`.** `==` compares values ("do these hold the same number?").
> `is` compares identity ("are these the exact same object in memory?"). For
> numbers and strings you almost always want `==`. The main place `is` matters:
> `x is None` -- use `is`, not `==`, when checking for `None`.

## Common questions and confusions

**"I wrote `if voltage = -55` and got a SyntaxError."**
That uses one `=`, which is assignment. Python does not allow assignment
inside an `if` condition (in most cases). You want `==`: `if voltage == -55`.

**"Why does `0.1 + 0.2 == 0.3` return `False`?"**
Computers store decimals in binary, and `0.1` has no exact binary form. The
stored value is slightly off, so the sum is `0.30000000000000004`, not `0.3`.
Never test two floats for exact equality; test whether they are close enough.
You will get a proper tool for this in the NumPy notebook.

**"Is `True` the same as `1`?"**
Yes, behind the scenes. `True == 1` returns `True`, and you can add booleans
as if they were numbers. This is intentional and useful for counting events.

**"Does order matter in `and`/`or`?"**
The final result is the same, but short-circuit evaluation means the left side
is checked first and the right may be skipped. For safety checks, put the
guard condition on the left.

**"What does `not not x` mean?"**
Double negation: `not not x` converts any value to its boolean equivalent.
`not not 5` is `True`; `not not 0` is `False`. It is an occasional idiom but
rarely the clearest choice.

## The idea: code that chooses a path

**Traffic lights.** When you drive, you do not apply the same action regardless
of the light colour. Red means stop. Amber means prepare. Green means go. The
light's colour selects which action runs. A conditional in Python works exactly
the same way: a condition selects which block of code runs.

**Triage in A&E.** A triage nurse does not give every patient the same
response. She checks vitals, classifies urgency, and routes the patient to the
appropriate bay -- resus, urgent, or minors. Each classification triggers a
different set of actions. That branching logic is a conditional.

In Python the keyword is `if`. When the condition is `True`, the indented block
below it runs. When it is `False`, Python skips that block entirely.

**Indentation is not decoration.** Python uses the amount of leading whitespace
(always four spaces, or one press of Tab) to know which lines belong to which
block. This is called **significant whitespace**. Get the indentation wrong and
the program either crashes or silently does the wrong thing.


````
# Simplest form: one decision, one outcome
voltage = -48.0
threshold = -55.0   # the voltage a neuron must exceed to fire a spike

if voltage > threshold:
    # This block runs ONLY when the condition is True
    print("Spike fired!")   # executes because -48 > -55

# Code here is outside the if block -- it always runs
print("Check complete.")
````

## Adding alternatives: elif and else

A single `if` handles "do this if yes." But decisions usually have more than
two outcomes. The A&E triage nurse doesn't just ask "is the patient critical?"
She asks a chain: critical? -- urgent? -- minor? Each question is checked in
order and exactly one route is taken.

**`elif`** (short for "else if") adds another condition to check only when all
previous conditions were `False`.

**`else`** catches everything that made it past all the conditions above. It
has no condition of its own; it simply says "if nothing above matched, do this."

Rules:
- One `if` (always first).
- Zero or more `elif` blocks (checked top to bottom; first one that is `True`
  wins and the rest are skipped).
- Zero or one `else` block (always last, if present).


````
# Classify a neuron's firing rate
# Firing rate (Hz) = spikes per second. A typical cortical neuron fires 1-100 Hz at rest.

firing_rate = 45.0   # Hz

if firing_rate < 10:
    category = "low"
elif firing_rate < 50:
    category = "medium"   # 45 falls here: < 50, so this block runs
elif firing_rate < 100:
    category = "high"
else:
    category = "very high / saturated"

print("Firing category:", category)   # medium

# The key insight: the checks run top-to-bottom and STOP at the first True one.
# Once 'medium' matches, Python does not even look at the 'high' branch.
````

## Nested conditionals: decisions inside decisions

Sometimes a decision only makes sense after a prior decision. A clinician
checks "is the patient in respiratory distress?" and only if yes asks
"is the SpO2 below 90%?" This is a **nested conditional**: an `if` block
inside another `if` block.

**The door lock.** The outer condition: "Is my keycard valid?" If yes, inner
condition: "Is the room unlocked at this time?" Both gates must be passed.


````
# Neuron model: check threshold, then classify spike amplitude
voltage = -47.0
threshold = -55.0
spike_amplitude = 80.0   # mV -- how large the electrical pulse is

if voltage > threshold:
    # We are inside the "neuron fired" branch
    print("Spike detected.")
    if spike_amplitude > 70:
        print("  Large spike -- strong input.")
    else:
        print("  Small spike -- weak input.")
else:
    # Neuron did not fire; nothing to classify
    print("No spike. Voltage:", voltage, "mV")
````

## Common questions and confusions

**"I got `IndentationError: expected an indented block`."**
You wrote `if ...:` but put nothing (or a comment with no code) inside the
block. Python requires at least one statement. Use `pass` as a placeholder:
`if condition: pass`.

**"My `else` never runs even when I expect it to."**
Check that the `else` lines up exactly with the `if` it belongs to. If it is
indented under an `elif`, it belongs to the `elif`, not the `if`.

**"Can I have multiple `else` blocks?"**
No. Only one `else` per `if` chain, and it must be last. For multiple
alternatives, use `elif`.

**"What if two `elif` conditions are both True?"**
Only the **first** one that is `True` runs. Python stops checking as soon as
it finds a match. The order of your conditions matters.

**"Do I need an `else`?"**
No. If nothing matches and there is no `else`, Python simply does nothing and
moves on. This is valid and common.

## The idea: doing the same thing to each item in a collection

**Checking items on a shopping list.** You hold a list and go through it one
item at a time: milk -- tick, bread -- tick, eggs -- tick. The action (check)
is the same every time; what changes is which item you are looking at.

**A ward round.** A consultant visits every patient in the bay, one after
another, and for each one checks observations, reviews notes, and writes a plan.
The procedure is identical for each patient; the *patient* changes.

A **for loop** in Python does exactly this: it takes a collection of items,
visits each one in turn, and runs the same block of code on it. The name you
give to "the current item" is the **loop variable** -- it gets a new value at
the start of each pass.

```
for loop_variable in collection:
    block that runs once per item
```

The colon and indentation work exactly as they do with `if`.


````
# Ward round: greet each patient
patients = ["Ali", "Beatrice", "Carlos", "Diana"]

for patient in patients:
    # 'patient' is a new variable that Python updates on every iteration
    print("Good morning,", patient)

# After the loop, 'patient' still exists and holds the LAST value
print("Last patient seen:", patient)
````

````
# Comp-neuro: sum spike counts across a population of neurons
# Each number is how many times that neuron fired in one second
spike_counts = [12, 0, 5, 22, 3, 17, 0, 9]

total = 0  # start the running total at zero

for count in spike_counts:
    total = total + count  # add this neuron's count to the running total

print("Total spikes across population:", total)
print("Mean firing rate:", total / len(spike_counts), "Hz")
# len() returns the number of items in the list
````

## range(): generating a sequence of numbers

Often you do not have a list to loop over -- you just want to repeat something
N times, or iterate over indices. `range()` generates a sequence of whole
numbers on demand (without storing them all in memory at once).

**The form you will use most:** `range(stop)` -- integers from 0 up to but not
including `stop`. So `range(5)` gives 0, 1, 2, 3, 4.

**Why does it start at 0?** Python counts from zero by convention (inherited
from C and carried everywhere). It feels odd at first but becomes automatic
within a week. A list of 5 items has indices 0, 1, 2, 3, 4.

Three forms of `range()`:

| Form | What it generates |
|------|------------------|
| `range(stop)` | 0, 1, ..., stop-1 |
| `range(start, stop)` | start, start+1, ..., stop-1 |
| `range(start, stop, step)` | start, start+step, ..., up to stop |


````
# range(stop): 0 up to but not including stop
print("range(5):", list(range(5)))   # [0, 1, 2, 3, 4]
# list() converts range to a visible list -- only for inspection; don't do this in production code

# range(start, stop)
print("range(2, 6):", list(range(2, 6)))   # [2, 3, 4, 5]

# range(start, stop, step): every second number
print("range(0, 10, 2):", list(range(0, 10, 2)))  # [0, 2, 4, 6, 8]

# Countdown: negative step
print("range(5, 0, -1):", list(range(5, 0, -1)))  # [5, 4, 3, 2, 1]
````

````
# Simulating time steps -- a core pattern in computational neuroscience
# Imagine a 10 ms recording sampled every 1 ms

dt = 1.0    # time step in milliseconds
total_time = 10   # total duration in ms

print("Simulating membrane voltage over time:")
voltage = -70.0   # start at resting voltage (mV)

for t in range(total_time):
    # At each time step, apply a small constant input current that raises voltage
    voltage = voltage + 0.8   # simplified: current nudges voltage up each ms
    print(f"t = {t} ms, voltage = {voltage:.1f} mV")
    # f-strings (f"...") will be covered fully in Notebook 3;
    # {voltage:.1f} means "show voltage rounded to 1 decimal place"
````

## Looping over a string

A string is a sequence of characters, and `for` loops treat it as such. This
is less common in data science but useful for understanding how sequences work.


````
region_name = "cortex"

for character in region_name:
    print(character)   # prints each letter on its own line
````

## Common questions and confusions

**"My loop variable seems wrong after the loop."**
The loop variable holds the **last** value it was assigned -- the final item
in the sequence. This is valid Python, just often surprising.

**"Why does `range(10)` give 0 to 9 and not 1 to 10?"**
Zero-based counting. The argument is a *count*, not an endpoint. `range(10)`
gives 10 items starting from 0.

**"Can I change the list I am looping over inside the loop?"**
You can, but it is dangerous and leads to confusing bugs. The safe rule: do not
modify a list while iterating over it. Build a new list or use a list
comprehension instead.

**"How do I loop backwards?"**
Use a negative step: `range(9, -1, -1)` gives 9, 8, ..., 0. Or use
`reversed(my_list)` for a list.

**"I get `NameError` on the loop variable after an empty loop."**
If the sequence was empty (e.g., `range(0)` or `[]`), the loop body never ran
and the loop variable was never created. Guard with a length check if needed.

# Item 4 — While Loops

## The idea: repeat until something changes

**The waiting room.** A patient sits and waits. The process is not "wait five
minutes" -- it is "keep waiting *until* your name is called." The number of
minutes is unknown in advance. The stopping condition is the event.

**An IV drip.** The nurse does not set a timer for "give 200 ml." She hangs
the bag and leaves it to run *while the bag is not empty*. The loop stops when
the condition changes.

A **`for` loop** is the right tool when you know the collection or count in
advance. A **`while` loop** is the right tool when you want to keep going
*until something becomes true* -- and you do not know when that will be.

```
while condition:
    block that repeats as long as condition is True
```

Each time the loop body finishes, Python checks the condition again. If it is
still `True`, the body runs again. If it is `False`, Python exits the loop.


````
# Simulate membrane voltage rising until it crosses the firing threshold
# This is a core pattern in integrate-and-fire neuron models.
# An integrate-and-fire model is the simplest mathematical description of
# how a neuron accumulates input and fires when voltage crosses a threshold.

voltage = -70.0      # starting resting voltage (mV)
threshold = -55.0    # firing threshold (mV)
input_current = 2.0  # constant current input per ms (mV per step)

step = 0  # count how many time steps we took

while voltage < threshold:
    voltage = voltage + input_current  # current nudges voltage up each step
    step += 1                          # += 1 is shorthand for step = step + 1
    print(f"Step {step}: voltage = {voltage:.1f} mV")

# When we reach here, voltage >= threshold: the neuron has fired
print(f"Spike at step {step}! Final voltage: {voltage:.1f} mV")
````

## break: exit immediately

`break` stops the loop right now, regardless of what the condition says.
It is like pressing the emergency stop. In neuroscience models you might
use `break` if voltage exceeds a maximum allowed value (a sanity check to
prevent a runaway simulation).


````
# Accumulate evidence until a decision threshold is reached, but cap at 20 steps
# This is loosely based on drift-diffusion models of decision-making --
# a model that says a brain "accumulates evidence" until it reaches a threshold
# and then commits to a choice.

import random
random.seed(42)   # fix randomness so the result is reproducible

evidence = 0.0
decision_threshold = 5.0
max_steps = 20

step = 0
while evidence < decision_threshold:
    sample = random.uniform(-1, 2)  # random evidence sample, slightly positive on average
    evidence += sample              # += adds to the running total
    step += 1
    print(f"Step {step}: evidence = {evidence:.2f}")

    if step >= max_steps:
        print("Maximum steps reached -- no decision made.")
        break
else:
    # The while...else block runs ONLY if the loop ended normally (not via break)
    print(f"Decision reached at step {step}! Evidence = {evidence:.2f}")
````

## continue: skip the rest of this iteration

`continue` jumps straight back to the condition check, skipping the remaining
lines in the current loop body. Think of it as "never mind this one, check the
next."

**Analogy:** A ward pharmacist checking prescriptions skips items that have
already been dispensed -- she sees the "dispensed" stamp and moves on without
re-reading the entry.


````
# Process a list of recorded voltages, but skip any marked as artifact (-999)
# Artifacts are spurious electrical readings caused by movement or equipment noise

voltages = [-70.0, -65.0, -999.0, -58.0, -999.0, -52.0, -48.0]
valid_readings = []

i = 0
while i < len(voltages):
    v = voltages[i]
    i += 1   # always advance the index, BEFORE the continue, or you loop forever

    if v == -999.0:
        print(f"Skipping artifact at position {i - 1}")
        continue   # jump back to while condition; do not append this value

    valid_readings.append(v)
    print(f"Accepted voltage: {v} mV")

print("Valid readings:", valid_readings)
````

## The infinite loop danger

A `while` loop whose condition never becomes `False` runs forever. This is an
**infinite loop**. It freezes your program (and eventually crashes your kernel
in a Jupyter notebook).

**The drip analogy gone wrong.** Imagine a drip that replenishes the bag each
time it empties. It runs forever. The stopping condition never arrives.

The two common causes:
1. You forgot to update the variable that the condition depends on.
2. You update it in the wrong direction (e.g., you meant to count up but
   counted down away from the threshold).

**Safeguard pattern:** always include a maximum iteration counter when writing
a new while loop you are not 100% sure about.

```
max_iter = 10000
iter_count = 0
while condition:
    ...
    iter_count += 1
    if iter_count > max_iter:
        raise RuntimeError("Loop exceeded maximum iterations -- check your logic.")
```


---
> ## Going deeper (optional on a first pass)
>
> **`while True:` with `break`.** A common pattern is an intentionally infinite
> loop that you exit with `break`. This is standard for "keep reading input
> until the user signals done." The structure is explicit about intent: the
> loop is designed to run until an internal condition triggers `break`.
>
> **`while ... else`.** Python's `while` loop has an optional `else` block that
> runs when the loop ends **normally** (condition became False) but NOT when it
> ends via `break`. You saw this in the evidence accumulation example above.
> It is rarely seen in the wild but elegant when it fits.
>
> **`for` vs `while` decision rule.** Use `for` when the sequence is known.
> Use `while` when you are waiting for an event or condition to change.
> In practice, a simulation that runs for exactly 1000 ms uses `for`;
> a simulation that runs until the model reaches steady state uses `while`.
>
> **Stochastic simulations.** Many neuroscience models involve randomness
> (noise in ion channels, variability in synaptic inputs). A while loop that
> stops when a threshold is crossed will terminate at a different step every
> run. Setting `random.seed(n)` before the loop fixes the random sequence so
> results are reproducible -- essential for debugging and for sharing results
> others can replicate.

## Common questions and confusions

**"My loop never starts."**
The condition was already `False` before the first iteration. A `while` loop
checks the condition *before* running the body. If the condition starts false,
the body never executes.

**"My loop runs forever."**
The variable driving the condition is not being updated, or is being updated
in the wrong direction. Add a print statement inside the loop to watch the
variable change (or not change).

**"When should I use `for` vs `while`?"**
Use `for` when you have a collection to iterate or a known count. Use `while`
when you are waiting for a condition to flip -- especially in simulations where
you do not know how many steps are needed.

**"Does `break` exit all loops or just the current one?"**
Just the **innermost** loop that contains it. In nested loops, you need
multiple `break` statements (or other logic) to exit all levels.

**"What happens to variables created inside the loop after it ends?"**
They persist. Python does not delete loop-local variables when the loop exits.
This differs from many other languages.

## enumerate(): index and value together

**The numbered handout.** A lecturer hands out numbered sheets: sheet 1 to
the first student, sheet 2 to the second. You know both *who* got the sheet
and *which sheet* they got. Without numbers you only know who; with numbers
you know where each person sits in the sequence.

**Marking scripts.** You mark a pile of exam scripts. You keep a running count
(script 1, script 2...) so you can say "candidate 3 missed question 4." The
count is the index; the candidate is the value.

`enumerate(collection)` wraps a collection and, on each iteration, hands you
a pair: `(index, value)`. You unpack the pair into two loop variables.


````
neuron_labels = ["V1", "M1", "PFC", "Hipp", "AMY"]
# Brain region abbreviations:
# V1 = primary visual cortex, M1 = primary motor cortex,
# PFC = prefrontal cortex, Hipp = hippocampus, AMY = amygdala

# Without enumerate: clunky index management
for i in range(len(neuron_labels)):
    print(i, neuron_labels[i])

print("---")

# With enumerate: clean and readable
for i, label in enumerate(neuron_labels):
    # 'i' is the position (0-based); 'label' is the item
    print(f"Neuron {i}: region {label}")
````

````
# enumerate with a custom start index
# In a multi-electrode array, electrodes are often numbered from 1, not 0
electrode_readings = [2.3, 0.0, 5.1, 4.7, 0.1]

for electrode_num, reading in enumerate(electrode_readings, start=1):
    status = "active" if reading > 1.0 else "silent"
    print(f"Electrode {electrode_num}: {reading} uV -- {status}")
````

## zip(): walking two lists side by side

**The patient wristband and the drug chart.** Two lists, both about the same
set of patients: one holds names, one holds doses. You need to read them
together: "Patient Ali -- 5 mg." `zip()` locks two lists together so you can
iterate both at once.

**Matching a key to a lock.** You have a bunch of keys and a bunch of locks,
in the same order. `zip()` hands you one key-lock pair at a time so you can
try them without losing track of which goes with which.

`zip(list_a, list_b)` produces pairs `(item_from_a, item_from_b)` in order.
You unpack each pair into two variables in the loop header.


````
neuron_ids  = [101, 102, 103, 104, 105]
firing_rates = [12.5, 0.0, 47.3, 88.1, 3.2]   # Hz: spikes per second

for neuron_id, rate in zip(neuron_ids, firing_rates):
    # Both lists are advanced in lockstep -- same index, both lists
    if rate > 0:
        print(f"Neuron {neuron_id} fires at {rate:.1f} Hz")
    else:
        print(f"Neuron {neuron_id} is silent")
````

````
# zip() with THREE lists
region_names  = ["V1",  "M1",  "PFC"]
mean_rates    = [14.2,  33.7,   8.1]   # Hz
peak_voltages = [-42.0, -38.5, -49.1]  # mV

for region, rate, peak in zip(region_names, mean_rates, peak_voltages):
    print(f"{region}: mean rate = {rate} Hz, peak voltage = {peak} mV")
````

## Combining enumerate() and zip()

You can nest these functions. `enumerate(zip(a, b))` gives you an index and a
pair on each iteration. This is a powerful pattern but read it carefully.


````
spike_times_neuron_A = [10, 25, 40, 58]   # ms: time of each spike
spike_times_neuron_B = [12, 26, 41, 59]   # ms

print("Paired spikes (potential synchrony check):")
for i, (t_A, t_B) in enumerate(zip(spike_times_neuron_A, spike_times_neuron_B)):
    # Synchrony: two neurons are considered synchronous if they fire within 5 ms of each other
    diff = abs(t_A - t_B)   # abs() gives the absolute (non-negative) difference
    sync = "synchronous" if diff <= 5 else "independent"
    print(f"  Spike pair {i}: A at {t_A} ms, B at {t_B} ms -- diff = {diff} ms -- {sync}")
````

---
> ## Going deeper (optional on a first pass)
>
> **What `zip` returns.** Like `range`, `zip` is lazy -- it generates pairs
> one at a time without storing the whole result. Wrap in `list()` to see it:
> `list(zip([1,2,3], ['a','b','c']))` gives `[(1, 'a'), (2, 'b'), (3, 'c')]`.
>
> **Unequal lengths.** If the two lists passed to `zip` have different lengths,
> `zip` stops at the shorter one silently. If you want to keep going and fill
> in a default value for the shorter list, use `itertools.zip_longest()` from
> the standard library.
>
> **Unzipping.** `zip(*pairs)` unpacks a list of pairs back into separate
> sequences. If `pairs = [(1, 'a'), (2, 'b')]`, then
> `ids, labels = zip(*pairs)` gives `ids = (1, 2)` and `labels = ('a', 'b')`.
> This is the reverse operation.
>
> **`enumerate` is also lazy.** It wraps any iterable, not just lists. You
> can enumerate a `range`, a `zip`, a generator, or a file object line by line.
>
> **Efficiency note.** `enumerate` and `zip` replace the verbose and
> error-prone pattern `for i in range(len(my_list)): val = my_list[i]`.
> Prefer them whenever you need index + value or parallel iteration. Modern
> Python style (called "Pythonic") strongly favours them.

## Common questions and confusions

**"I get a tuple in my loop variable, not two separate things."**
You probably wrote `for item in enumerate(my_list):` instead of
`for i, item in enumerate(my_list):`. The parentheses in the loop header
unpack the pair. Without them, `item` is the whole `(index, value)` tuple.

**"zip stopped early and I lost data."**
One of your lists is shorter than the other. Check lengths with
`len(list_a)` and `len(list_b)` before zipping.

**"Can I enumerate and zip at the same time?"**
Yes: `enumerate(zip(a, b))`. The loop header looks like
`for i, (val_a, val_b) in enumerate(zip(a, b)):`. The inner parentheses in
the header are essential to unpack the pair that zip produces.

**"Does `enumerate` work on strings?"**
Yes. Any sequence. `enumerate("abc")` gives `(0, 'a'), (1, 'b'), (2, 'c')`.

````
# Solution: Exercise 6 (stretch)
voltage = -70.0
threshold = -55.0
input_per_step = 1.5
spike_count = 0
step = 0

while spike_count < 3:
    voltage += input_per_step
    step += 1
    if voltage >= threshold:
        spike_count += 1
        print(f"SPIKE {spike_count} at step {step}! Resetting voltage.")
        voltage = -70.0   # reset to resting potential after firing
````

````
# Solution: Exercise 4
firing_rates = [12.0, 45.0, 3.5, 88.2, 22.1]
max_rate = firing_rates[0]
max_index = 0

for i, rate in enumerate(firing_rates):
    if rate > max_rate:
        max_rate = rate
        max_index = i

print(f"Max firing rate: {max_rate} Hz at index {max_index}")   # 88.2 at index 3
````
